# Stochastic Gradient Descent

Stochastic Gradient Descent (SGD) is an optimization algorithm used to train machine learning models — especially large-scale models like neural networks — by updating the model parameters step-by-step to minimize a loss function.

**In Stochastic Gradient Descent:**

- Instead of computing the gradient using the entire dataset (which can be slow for large datasets), we use only one sample (or a very small batch) at each update.
- This makes updates faster and introduces some noise in the updates, which can help escape local minima.

## Difference from Other Gradient Descent Variants

| Method                     | Data Used Per Update          | Pros                          | Cons                         |
| -------------------------- | ----------------------------- | ----------------------------- | ---------------------------- |
| **Batch Gradient Descent** | Entire dataset                | Stable, accurate updates      | Very slow for large data     |
| **Stochastic GD**          | 1 sample                      | Fast, can escape local minima | Updates noisy, may oscillate |
| **Mini-batch GD**          | Small subset (10–512 samples) | Balanced speed & stability    | Needs tuning of batch size   |


```python
from sklearn.datasets import load_diabetes

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
reg = LinearRegression()
reg.fit(X_train,y_train)
```

```python
print(reg.coef_)
print(reg.intercept_)
```

```python
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

## we define the above to know the the coef_ and intercept_

```python
class SGDRegressor:

    def __init__(self,lr=0.01,epochs=100):
        self.lr=lr
        self.epochs=epochs
        self.intercept_=None
        self.coef_=None

    def fit(self,X_train,y_train):
        self.intercept_=0
        self.coef_=np.ones(X_train.shape)[1] # to define 10 coef_  as 1

        for i in range(self.epochs):
            for j in range(X_train.shape[0]):
                idx=np.random.randint(0,X_train.shape[0])

                y_hat=np.dot(X_train[idx],self.coef_)+self.intercept_  ## sacler output

                intercept_der= -2 * (y_train[idx] -y_hat) 
                self.intercept_= self.intercept_ -(self.lr * intercept_der)

                coef_der= -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                self.coef_= self.coef_ - (self.lr * coef_der)
            


        
        print(self.intercept_,self.coef_)
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
sgd=SGDRegressor(lr=0.01,epochs=500)
```

```python
sgd.fit(X_train,y_train)
```

```python
y_pred=sgd.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
