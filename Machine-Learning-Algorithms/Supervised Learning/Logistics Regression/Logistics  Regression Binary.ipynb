```python
df=pd.DataFrame(data=data.data,columns=(['sepal length (cm)',
 'sepal width (cm)',
 'petal length (cm)',
 'petal width (cm)']))
```

```python
 ## **an other way**
df=pd.DataFrame(data.data,columns=data.feature_names)   
```

```python
df
```

## Initial  Binary classfication   

so we will take only 2 outcome **(0,1)**


```python
df["target"]=data.target
```

```python
df
```

```python
data.target
```

```python
## Binary classfication 

df=df[df["target"]!=2]
```

```python
df
```

 ##  Independent and Dependent fetaures ** Binary**

```python
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
X
```

```python
y
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=20,random_state=42)
```

## LogisticRegression

```python
from sklearn.linear_model import LogisticRegression
```

```python
classifier=LogisticRegression() 
```

```python
classifier.fit(X_train,y_train)
```

```python
classifier.predict_proba(X_test)
```

## prediction

```python
y_pred=classifier.predict(X_test)
```

```python
y_pred
```

## Confusion matrix,accuracy score ,classification report

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
print(confusion_matrix(y_pred,y_test))
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```

## Hyperparameter Tuning 

**GridseachCV**

```python
from sklearn.model_selection import GridSearchCV
import warnings
warnings.filterwarnings("ignore")
```

```python
parameter={"penalty":("l2","l1","elasticnet",None),"C":[1,10,20]}
```

```python
clf=GridSearchCV(classifier,param_grid=parameter,cv=5)
```

```python
clf.fit(X_train,y_train)
```

```python
clf.best_params_  ## best parameters
```

```python

classifier=LogisticRegression(C=1,penalty='l2')
```

```python
classifier.fit(X_train,y_train)
```

```python
y_pred=classifier.predict(X_test)
```

```python
print(confusion_matrix(y_pred,y_test))
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```

**RandomizedSerchCV**

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
random_clf=RandomizedSearchCV(classifier,param_distributions=parameter,cv=5)
```

```python
random_clf.fit(X_train,y_train)
```

```python
random_clf.best_params_
```

```python
classifier=LogisticRegression(penalty=None,C=10)
```

```python
classifier.fit(X_train,y_train)
```

```python
y_pred=classifier.predict(X_test)
```
