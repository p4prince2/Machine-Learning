```python
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
```

```python
df
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
df1=OneHotEncoder()
```

```python
df1
```

```python
df3=df1.fit_transform(df[['day','sex','smoker']]).toarray()  
```

```python
import pandas as pd
```

```python
X=pd.DataFrame(df3,columns=df1.get_feature_names_out())
```

```python
X
```

```python
import numpy as np
```

```python
k=np.where(df['time'] == 'Dinner', 1, 0)
```

```python
k
```

```python
y=pd.DataFrame(k)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,x_test,y_train,y_test=train_test_split(X,k,random_state=0,test_size=.2)
```

```python
from sklearn.naive_bayes import BernoulliNB
```

```python
nb=BernoulliNB()
```

```python
nb.fit(X_train,y_train)
```

```python
y_pred=nb.predict(x_test)
```

```python
from sklearn.metrics import r2_score,classification_report,accuracy_score,confusion_matrix
```

```python
print(confusion_matrix(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
