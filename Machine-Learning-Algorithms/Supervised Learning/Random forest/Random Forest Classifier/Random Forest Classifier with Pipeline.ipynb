```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.impute import SimpleImputer  ## Handle missing value
from sklearn.preprocessing import StandardScaler  ## feature scaling
from sklearn.preprocessing  import OneHotEncoder ## categorical to numerical
from sklearn.compose import ColumnTransformer
```

```python
categorical_cols=['sex','smoker','day']
numerical_cols=['total_bill','tip','size']
```

```python
##Feature Engineering Automation
## Numerical Pipelines 
num_pipline= Pipeline(
steps=[
     ('imputer',SimpleImputer(strategy='median')),## missing values
    ('scaler',StandardScaler()) ## feature scaling
    
]
)


## Categorical Piplelines

cat_pipline=Pipeline(
    steps=[
         ('imputer',SimpleImputer(strategy='most_frequent')), ## missinf value
        ('onehotencoder',OneHotEncoder())  ## categorical feature to numerical feature
        
    ]
)
```

```python
preprocessor=ColumnTransformer(
    [
        ('num_pipeline',num_pipline,numerical_cols),
        ('cat_pipeline',cat_pipline,categorical_cols)
    ]
) 
```

```python


 X_train=preprocessor.fit_transform(X_train)

 X_test=preprocessor.fit_transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
## Model Training Automation

models={
    'Random Forest':RandomForestClassifier(),
    'Logistic Regression':LogisticRegression()
    ,'Decision Tree':DecisionTreeClassifier()
}
```

```python
from sklearn.metrics import accuracy_score
```

```python
def evaluate_moel(X_train,y_train,X_test,y_test,models):
    report={}
    for i in range(len(models)):
        model=list(models.values())[i]
        #Train Model
        model.fit(X_train,y_train)


        #Predict Testing Data
        y_test_pred=model.predict(X_test)
        
        #Get R2 scores for  test data
        test_model_score=accuracy_score(y_test,y_test_pred)
        report[list(models.keys())[i]]=test_model_score

    return report
```

```python
evaluate_moel(X_train,y_train,X_test,y_test,models)
```

## ** Random forest **  gives the best result

## Hyperparameter Tuning

```python
classifier=RandomForestClassifier()
```

```python
para={
    'max_depth':[3,5,10,None],
    'n_estimators':[100,200,300]## Number of decision tree to take
    ,'criterion':['gini','entropy']
    
}
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
gridCV=RandomizedSearchCV(classifier,param_distributions=para,verbose=3,cv=5,scoring='accuracy')
gridCV.fit(X_train,y_train)
```

```python
gridCV.best_params_
```

```python
y_pred=gridCV.predict(X_test)
```

```python
accuracy_score(y_test,y_pred)
```
