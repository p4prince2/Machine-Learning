```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_breast_cancer

```

```python
data=load_breast_cancer()
```

```python
data
```

```python
print(data.DESCR)
```

```python
data.target
```

```python
data.feature_names
```

```python
df=pd.DataFrame(data=data.data,columns=data.feature_names)
```

```python
df
```

```python
X=df.iloc[::]
y=data.target
```

```python
X,y
```

```python
from sklearn.model_selection import train_test_split
X_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.tree import DecisionTreeClassifier
desc=DecisionTreeClassifier()
desc.fit(X_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(20,20))
tree.plot_tree(desc,filled=True)
```

Max Depth =2

```python
dec_max=DecisionTreeClassifier(max_depth=3)
dec_max.fit(X_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(dec_max,filled=True)
```

```python
y_pred=dec_max.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
print(accuracy_score(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
print(classification_report(y_pred,y_test))
```
