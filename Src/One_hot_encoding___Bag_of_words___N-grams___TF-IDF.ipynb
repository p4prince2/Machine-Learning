```python
# Bag of Words | TF-IDF | N-grams (Uni-gram + Bi-gram)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
data = {
    "text": [
        "people watch campusx",
        "campusx watch campusx",
        "people write comment",
        "campusx write comment"
    ],
    "output": [1, 1, 0, 0]
}

df = pd.DataFrame(data)
df

```

# One Hot Encoding

```python
# binary=True makes values 0 or 1 only
one_hot = CountVectorizer(binary=True)

X_onehot = one_hot.fit_transform(df["text"])

onehot_df = pd.DataFrame(
    X_onehot.toarray(),
    columns=one_hot.get_feature_names_out()
)

print(onehot_df)
print(df)
```

# Bag of Words (BoW)

```python
bow = CountVectorizer()

X_bow = bow.fit_transform(df["text"])

bow_df = pd.DataFrame(
    X_bow.toarray(),
    columns=bow.get_feature_names_out()
)

print(bow_df)
```

# Uni-gram (1-word sequence)

```python
unigram = CountVectorizer(ngram_range=(1, 1))

X_uni = unigram.fit_transform(df["text"])

unigram_df = pd.DataFrame(
    X_uni.toarray(),
    columns=unigram.get_feature_names_out()
)

print(unigram_df)

```

# Bi-gram (2-word sequence)

```python
bigram = CountVectorizer(ngram_range=(2, 2))

X_bi = bigram.fit_transform(df["text"])

bigram_df = pd.DataFrame(
    X_bi.toarray(),
    columns=bigram.get_feature_names_out()
)

print(bigram_df)
```

# N-gram (Uni-gram + Bi-gram)

```python
print("\n========== N-gram (1,2) ==========\n")

ngram = CountVectorizer(ngram_range=(1, 2))

X_ngram = ngram.fit_transform(df["text"])

ngram_df = pd.DataFrame(
    X_ngram.toarray(),
    columns=ngram.get_feature_names_out()
)

print(ngram_df)
```

# TF-IDF

```python
tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(df["text"])

tfidf_df = pd.DataFrame(
    X_tfidf.toarray(),
    columns=tfidf.get_feature_names_out()
)

print(tfidf_df)
```
