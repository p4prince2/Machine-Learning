```python
import numpy as np
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()
X = iris.data  # Data matrix X

# Calculate the mean vector
X_mean = np.mean(X, axis=0)

X_center = X - X_mean

print(X_center.shape)
```

```python
import numpy as np

U,S,V_t = np.linalg.svd(X_center)
```

```python
V_t
```

|         | f1     | f2     | f3     | f4     |
| ------: | ------ | ------ | ------ | ------ |
| **PC1** | 0.361  | -0.084 | 0.857  | 0.358  |
| **PC2** | -0.657 | -0.730 | 0.173  | 0.075  |
| **PC3** | 0.582  | -0.598 | -0.076 | -0.546 |
| **PC4** | 0.315  | -0.320 | -0.480 | 0.754  |


```python
V= V_t.T
```

```python
V
```

V =
[

        PC1    PC2     PC3    PC4
   
    [ 0.361  -0.657   0.582   0.315 ]   ← f1
 
    [-0.084  -0.730  -0.598  -0.320 ]   ← f2
 
    [ 0.857   0.173  -0.076  -0.480 ]   ← f3
 
    [ 0.358   0.075  -0.546   0.754 ]   ← f4

]


```python
V[:,0:2]
```

This is:

- 2 vectors (PC1, PC2)

- Each vector has 4 numbers

- Each vector is a direction in the same 4-D space

-  PC1 = [ 0.361, -0.084,  0.857,  0.358 ]
-  PC2 = [-0.657, -0.730,  0.173,  0.075 ]


####  To map 4-dimensional data into 2-dimensional

```python
X_transformed = np.dot(X_center,V[:,0:2])
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_transformed[:,0],X_transformed[:,1], c=iris.target)
```
